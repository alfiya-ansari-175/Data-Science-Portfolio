# 21. Cumulative titles added over years

```python
%%sql
WITH yearly_totals AS(
    select YEAR(STR_TO_DATE(date_added, '%M %d, %Y')) AS year_added, count(*) as total from table1
    where date_added IS NOT NULL
    group by year_added                                                                           
)

select year_added, SUM(total) OVER(ORDER BY year_added ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) as cumulative_titles
from yearly_totals
order by year_added;
```

# 30. Top 3 genres per year

```python
%%sql
WITH ranked_categories AS (
    select 
        release_year,listed_in, count(*) as total,
        DENSE_RANK() OVER (PARTITION BY release_year ORDER BY COUNT(*) DESC) AS rnk
    from table1
    group by release_year, listed_in
)
select release_year, listed_in, total
from ranked_categories
where rnk <= 3;
```

# 35. Average number of titles added per year

```python
%%sql
WITH yearly_totals AS(
    select YEAR(STR_TO_DATE(date_added, '%M %d, %Y')) as year,
    count(*) as total
    from table1
    group by year
)
select avg(total) as avg_titles_per_year from yearly_totals;
```

# 36. Is Netflix shifting focus from Movies to TV Shows?

```python
%%sql
select YEAR(STR_TO_DATE(date_added, '%M %d, %Y')) as year_added,
    sum(type='Movie') as movies,
    sum(type='TV Show') as tv_shows
from table1
where date_added is NOT NULL
group by year_added
order by year_added;
```

# 37. Are international productions increasing? (Not US-based)

```python
%%sql
select 
    YEAR(STR_TO_DATE(date_added, '%M %d, %Y')) AS year_added,
    count(*) AS international_titles
from table1
where country not like '%United States%'
  and date_added is not null
group by year_added
order by year_added;
```

# 38. What genres dominate different regions?

```python
%%sql
WITH country_genre AS (
    select country, listed_in, count(*) AS total,
        RANK() OVER (PARTITION BY country ORDER BY COUNT(*) DESC) AS rnk
    from table1
    group by country, listed_in
)
select country, listed_in AS top_genre, total
from country_genre
where rnk = 1
limit 10;
```
