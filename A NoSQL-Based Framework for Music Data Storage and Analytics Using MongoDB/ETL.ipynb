```python
from pymongo import MongoClient
import pandas as pd
```

```python
mongo_uri = "mongodb://localhost:27017/"
db_name = "SpotifyAnalytics"
collection_name = "tracks"
```

```python
client = MongoClient(mongo_uri)
db = client[db_name]
collection = db[collection_name]
```

```python
df = pd.read_csv('/Users/alfiyaansari/Desktop/Important/Project/spotify_2015_2025_85k.csv')
df.head()
```

```python
df['duration_ms'] = round(df['duration_ms'] / 60000)
df.rename(columns={"duration_ms" : "duration_minutes"},inplace=True)
```

```python
df.head()
```

```python
df['release_date'] = pd.to_datetime(df['release_date'], errors = 'coerce')
df['release_year'] = df['release_date'].dt.year
```

```python
def pop_level(x):
    if x >= 70:
        return 'High'
    elif x >= 40:
        return 'Medium'
    else:
        return 'Low'
```

```python
df['popularity_level'] = df['popularity'].apply(pop_level)
```

```python
df.head()
```

```python
pd.set_option('display.max_columns',50)
```

```python
df.head()
```

```python
records = df.to_dict(orient='records')
```

```python
collection.insert_many(records)
print(f'Inserted {len(records)} spotify track documents')
```
